```python
import numpy as np
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# 1. Environment & Setup
# ---------------------------------------------------------
# Define search space bounds [xmin, ymin], [xmax, ymax]
lb = np.array([0, 0])
ub = np.array([10, 10])

# Define start, goal, and circular obstacles [x, y, radius]
start = np.array([0.5, 0.5])
goal = np.array([9.5, 9.5])
obstacles = [
    np.array([3.0, 3.0, 1.2]),
    np.array([5.0, 7.0, 1.5]),
    np.array([7.0, 4.0, 1.0])
]

# ---------------------------------------------------------
# 2. Objective (Fitness) Function
# ---------------------------------------------------------
def fitness_function(position):
    # Constraint 1: Out of bounds check
    if np.any(position < lb) or np.any(position > ub):
        return 1e6

    # Constraint 2: Obstacle collision check
    for obs in obstacles:
        obs_pos = obs[:2]
        radius = obs[2]
        distance = np.linalg.norm(position - obs_pos)
        if distance <= radius:
            # Apply an aggressive mathematical penalty for collisions
            return 1e6 + (radius - distance) * 1000

    # Objective: Minimize standard Euclidean distance to the goal
    return np.linalg.norm(position - goal)

# ---------------------------------------------------------
# 3. Grey Wolf Optimisation (GWO) Core Loop
# ---------------------------------------------------------
def grey_wolf_optimization(num_wolves=30, max_iter=50):
    dim = 2
    # Initialize wolf positions randomly within the mapping framework
    Positions = np.random.uniform(lb, ub, (num_wolves, dim))

    # Pre-allocate hierarchy vector spaces
    Alpha_pos = np.zeros(dim)
    Beta_pos = np.zeros(dim)
    Delta_pos = np.zeros(dim)

    Alpha_score = float("inf")
    Beta_score = float("inf")
    Delta_score = float("inf")

    history = []

    for l in range(max_iter):
        for i in range(num_wolves):
            # Enforce map limits
            Positions[i] = np.clip(Positions[i], lb, ub)
            fitness = fitness_function(Positions[i])

            # Rank hierarchy assignments (Alpha, Beta, Delta)
            if fitness < Alpha_score:
                Delta_score = Beta_score
                Delta_pos = Beta_pos.copy()
                Beta_score = Alpha_score
                Beta_pos = Alpha_pos.copy()
                Alpha_score = fitness
                Alpha_pos = Positions[i].copy()

            elif fitness < Beta_score:
                Delta_score = Beta_score
                Delta_pos = Beta_pos.copy()
                Beta_score = fitness
                Beta_pos = Positions[i].copy()

            elif fitness < Delta_score:
                Delta_score = fitness
                Delta_pos = Positions[i].copy()

        # Linearly decrease parameter 'a' from 2 to 0 to shift from exploration to exploitation
        a = 2 - l * (2 / max_iter)

        # Compute hunting vectors for Omega pack units
        for i in range(num_wolves):
            # Alpha updates
            r1, r2 = np.random.rand(dim), np.random.rand(dim)
            A1 = 2 * a * r1 - a
            C1 = 2 * r2
            D_alpha = abs(C1 * Alpha_pos - Positions[i])
            X1 = Alpha_pos - A1 * D_alpha

            # Beta updates
            r1, r2 = np.random.rand(dim), np.random.rand(dim)
            A2 = 2 * a * r1 - a
            C2 = 2 * r2
            D_beta = abs(C2 * Beta_pos - Positions[i])
            X2 = Beta_pos - A2 * D_beta

            # Delta updates
            r1, r2 = np.random.rand(dim), np.random.rand(dim)
            A3 = 2 * a * r1 - a
            C3 = 2 * r2
            D_delta = abs(C3 * Delta_pos - Positions[i])
            X3 = Delta_pos - A3 * D_delta

            # Update position based on the mean trajectory of the three leaders
            Positions[i] = (X1 + X2 + X3) / 3

        history.append(Alpha_pos.copy())

    return Alpha_pos, history

# Execute GWO Strategy
best_waypoint, path_history = grey_wolf_optimization()

# ---------------------------------------------------------
# 4. Matplotlib Visualization Output
# ---------------------------------------------------------
fig, ax = plt.subplots(figsize=(8, 8))

# Draw obstacles
for obs in obstacles:
    circle = plt.Circle((obs[0], obs[1]), obs[2], color='crimson', alpha=0.5, label='Obstacle Penalty Zone' if obs[0]==3.0 else "")
    ax.add_patch(circle)

# Draw start and target coordinates
ax.scatter(start[0], start[1], color='blue', s=150, marker='^', label='Robot Start Configuration', zorder=5)
ax.scatter(goal[0], goal[1], color='darkgreen', s=200, marker='*', label='Goal Target Location', zorder=5)

# Trace trajectory history
path_history = np.array([start] + path_history)
ax.plot(path_history[:, 0], path_history[:, 1], color='darkorange', linestyle='--', marker='o', linewidth=2, label='GWO Calculated Safe Route')

ax.set_xlim(lb[0], ub[0])
ax.set_ylim(lb[1], ub[1])
ax.set_xlabel('X Coordinate (Meters)')
ax.set_ylabel('Y Coordinate (Meters)')
ax.set_title('Autonomous Robot Navigation via Grey Wolf Optimisation (GWO)')
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='upper left')
plt.show()
```
