```python
import numpy as np
import matplotlib.pyplot as plt


GRID_SIZE = 10
NUM_PARTICLES = 30
MAX_ITER = 50
BUDGET_LIMIT = 25


np.random.seed(42)
HEAT_MAP = np.zeros((GRID_SIZE, GRID_SIZE))
HEAT_MAP[3:7, 3:7] = 2.0


def evaluate_fitness(layout):
    """
    Evaluates how good a tree layout is.
    Higher score = Better layout.
    """
    cooling_benefit = np.sum(layout * HEAT_MAP)

    coverage = np.sum(layout)

    penalty = 0
    if coverage > BUDGET_LIMIT:
        penalty = (coverage - BUDGET_LIMIT) * 5.0

    fitness = cooling_benefit + (coverage * 0.1) - penalty
    return fitness

class Particle:
    def __init__(self):
        self.position = np.random.rand(GRID_SIZE, GRID_SIZE)
        self.velocity = np.random.uniform(-0.1, 0.1, (GRID_SIZE, GRID_SIZE))
        self.best_position = np.copy(self.position)
        self.best_fitness = -float('inf')

    def get_binary_layout(self):
        flat = self.position.flatten()
        idx = np.argsort(flat)[-BUDGET_LIMIT:]
        binary_flat = np.zeros(GRID_SIZE * GRID_SIZE)
        binary_flat[idx] = 1
        return binary_flat.reshape(GRID_SIZE, GRID_SIZE)

    def update(self, global_best_pos, w=0.5, c1=1.5, c2=1.5):
        r1, r2 = np.random.rand(GRID_SIZE, GRID_SIZE), np.random.rand(GRID_SIZE, GRID_SIZE)

        cognitive = c1 * r1 * (self.best_position - self.position)
        social = c2 * r2 * (global_best_pos - self.position)
        self.velocity = w * self.velocity + cognitive + social

        self.velocity = np.clip(self.velocity, -0.2, 0.2)
        self.position += self.velocity
        self.position = np.clip(self.position, 0.0, 1.0)

swarm = [Particle() for _ in range(NUM_PARTICLES)]
global_best_position = np.random.rand(GRID_SIZE, GRID_SIZE)
global_best_fitness = -float('inf')
history = []

for iteration in range(MAX_ITER):
    for particle in swarm:
        layout = particle.get_binary_layout()
        fitness = evaluate_fitness(layout)

        if fitness > particle.best_fitness:
            particle.best_fitness = fitness
            particle.best_position = np.copy(particle.position)

        if fitness > global_best_fitness:
            global_best_fitness = fitness
            global_best_position = np.copy(particle.position)

    history.append(global_best_fitness)

best_particle = Particle()
best_particle.position = global_best_position
optimized_layout = best_particle.get_binary_layout()

print(f"Optimization Complete! Best Fitness Score Achieved: {global_best_fitness:.2f}")
print(f"Total trees planted: {int(np.sum(optimized_layout))}")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(range(1, MAX_ITER + 1), history, color='forestgreen', linewidth=2)
axes[0].set_title('Algorithm Convergence (Optimization)')
axes[0].set_xlabel('Iteration')
axes[0].set_ylabel('Fitness Score (Higher is Better)')
axes[0].grid(True, linestyle='--')

im = axes[1].imshow(HEAT_MAP, cmap='YlOrRd', origin='lower', alpha=0.6)
y_indices, x_indices = np.where(optimized_layout == 1)
axes[1].scatter(x_indices, y_indices, color='darkgreen', s=180, marker='^', label='Optimized Tree Position')
axes[1].set_title('Final Optimized Tree Placement')
axes[1].set_xticks(range(GRID_SIZE))
axes[1].set_yticks(range(GRID_SIZE))
axes[1].grid(True, color='gray', linestyle=':', alpha=0.5)
axes[1].legend(loc='upper right')

plt.colorbar(im, ax=axes[1], label='Heat Intensity Value')
plt.tight_layout()
plt.show()
```
