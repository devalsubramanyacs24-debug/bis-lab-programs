```python
import random

STAFF = ['Nurse A', 'Nurse B', 'Nurse C', 'Nurse D']
SHIFTS = ['Morning', 'Evening', 'Night' , 'Off']
DAYS = 7

def initialize_population(pop_size, staff, shifts, days):
    population = []
    for _ in range(pop_size):
        schedule = []
        for day in range(days):
            daily_assignments = []
            assigned_staff_today = set()
            for shift in shifts:
                available_staff = [s for s in staff if s not in assigned_staff_today]
                if not available_staff:
                    available_staff = list(staff)
                chosen_staff = random.choice(available_staff)
                daily_assignments.append((day, shift, chosen_staff))
                assigned_staff_today.add(chosen_staff)
            schedule.extend(daily_assignments)
        population.append(schedule)
    return population

POPULATION_SIZE = 10
initial_population = initialize_population(POPULATION_SIZE, STAFF, SHIFTS, DAYS)
print(f"Initial Population Size: {len(initial_population)}")
print("First schedule example (first 5 assignments):")
print(initial_population[0][:5])
```
