```python


import math
import random
import numpy as np
import pandas as pd


NUM_COURSES = 12
NUM_SLOTS = 6
NUM_ROOMS = 3

ROOM_CAPACITIES = {
    0: 50,
    1: 40,
    2: 30
}

TOTAL_OPTIONS = NUM_SLOTS * NUM_ROOMS

def index_to_slot_room(idx):
    slot = idx // NUM_ROOMS
    room = idx % NUM_ROOMS
    return int(slot), int(room)

np.random.seed(42)
random.seed(42)

COURSE_ENROLLMENTS = {i: int(np.random.randint(15, 45)) for i in range(NUM_COURSES)}

conflict_matrix = np.zeros((NUM_COURSES, NUM_COURSES), dtype=int)
for i in range(NUM_COURSES):
    for j in range(i + 1, NUM_COURSES):
        if random.random() < 0.3:
            shared = int(random.randint(1, 15))
            conflict_matrix[i][j] = shared
            conflict_matrix[j][i] = shared


def calculate_fitness(nest_vector):
    """
    Evaluates penalty scores across constraints to establish localized fitness.
    Goal: Minimize student conflicts and room overruns.
    """
    conflicts = 0
    penalties = 0
    slot_room_occupancy = {}

    schedule = {}
    for course_id, option_idx in enumerate(nest_vector):
        schedule[course_id] = index_to_slot_room(int(option_idx))

    for course, (slot, room) in schedule.items():
        if (slot, room) not in slot_room_occupancy:
            slot_room_occupancy[(slot, room)] = []
        slot_room_occupancy[(slot, room)].append(course)

        enrollment = COURSE_ENROLLMENTS[course]
        capacity = ROOM_CAPACITIES[room]
        if enrollment > capacity:
            penalties += (enrollment - capacity) * 5

    for (slot, room), courses in slot_room_occupancy.items():
        if len(courses) > 1:
            conflicts += (len(courses) - 1) * 50

    for i in range(NUM_COURSES):
        for j in range(i + 1, NUM_COURSES):
            if schedule[i] == schedule[j]:
                shared_students = conflict_matrix[i][j]
                if shared_students > 0:
                    conflicts += shared_students * 10

    total_penalty = conflicts + penalties
    fitness = 1.0 / (total_penalty + 1)
    return fitness, total_penalty, conflicts, penalties


def levy_flight(Lambda=1.5):
    """
    Generates step sizes following a Lévy distribution using Mantegna's algorithm.
    """
    sigma_u = (math.gamma(1 + Lambda) * math.sin(math.pi * Lambda / 2) /
              (math.gamma((1 + Lambda) / 2) * Lambda * (2 ** ((Lambda - 1) / 2)))) ** (1 / Lambda)
    sigma_v = 1

    u = np.random.normal(0, sigma_u, NUM_COURSES)
    v = np.random.normal(0, sigma_v, NUM_COURSES)

    step = u / (np.abs(v) ** (1 / Lambda))
    return step


NUM_NESTS = 25
PA = 0.25
MAX_GENERATIONS = 100

nests = np.random.randint(0, TOTAL_OPTIONS, size=(NUM_NESTS, NUM_COURSES)).astype(float)
fitness_array = np.zeros(NUM_NESTS)
penalty_array = np.zeros(NUM_NESTS)

for i in range(NUM_NESTS):
    fit, pen, _, _ = calculate_fitness(nests[i])
    fitness_array[i] = fit
    penalty_array[i] = pen

best_idx = np.argmax(fitness_array)
best_nest = nests[best_idx].copy()
best_penalty = penalty_array[best_idx]

print("Starting Cuckoo Search for Exam Scheduling...\n")

for generation in range(MAX_GENERATIONS):

    random_nest_idx = random.randint(0, NUM_NESTS - 1)
    cuckoo = nests[random_nest_idx].copy()

    step_size = levy_flight()
    cuckoo += step_size * (cuckoo - best_nest)
    cuckoo = np.clip(cuckoo, 0, TOTAL_OPTIONS - 1)
    cuckoo_discrete = np.round(cuckoo).astype(int)

    cuckoo_fit, cuckoo_pen, _, _ = calculate_fitness(cuckoo_discrete)

    target_idx = random.randint(0, NUM_NESTS - 1)
    if cuckoo_fit > fitness_array[target_idx]:
        nests[target_idx] = cuckoo_discrete.copy()
        fitness_array[target_idx] = cuckoo_fit
        penalty_array[target_idx] = cuckoo_pen


    sorted_indices = np.argsort(fitness_array)
    num_abandoned = int(PA * NUM_NESTS)

    for i in range(num_abandoned):
        abandoned_idx = sorted_indices[i]


        n1, n2 = random.sample(range(NUM_NESTS), 2)
        step_factor = random.random()
        new_nest = nests[abandoned_idx] + step_factor * (nests[n1] - nests[n2])
        new_nest = np.clip(new_nest, 0, TOTAL_OPTIONS - 1)
        new_nest_discrete = np.round(new_nest).astype(int)

        new_fit, new_pen, _, _ = calculate_fitness(new_nest_discrete)
        nests[abandoned_idx] = new_nest_discrete
        fitness_array[abandoned_idx] = new_fit
        penalty_array[abandoned_idx] = new_pen

    current_best_idx = np.argmax(fitness_array)
    if penalty_array[current_best_idx] < best_penalty:
        best_penalty = penalty_array[current_best_idx]
        best_nest = nests[current_best_idx].copy()

    if (generation + 1) % 25 == 0 or generation == 0:
        avg_penalty = np.mean(penalty_array)
        print(f"Generation {generation+1:03d} | Best Penalty: {best_penalty:.1f} | Avg Penalty: {avg_penalty:.1f}")


print("\n" + "="*50)
print("OPTIMIZATION COMPLETE")
print("="*50)
_, final_penalty, final_conflicts, final_penalties = calculate_fitness(best_nest.astype(int))

print(f"Final Global Penalty Score: {final_penalty}")
print(f"-> Student & Room Overbooking Conflicts: {final_conflicts}")
print(f"-> Room Capacity Penalties: {final_penalties}")

parsed_schedule = []
for course_id, option_idx in enumerate(best_nest.astype(int)):
    slot, room = index_to_slot_room(option_idx)
    parsed_schedule.append({
        "Course ID": f"Course {course_id:02d}",
        "Time Slot": f"Slot {slot + 1}",
        "Assigned Room": f"Room {room} (Cap: {ROOM_CAPACITIES[room]})",
        "Enrollment": COURSE_ENROLLMENTS[course_id]
    })

df_res = pd.DataFrame(parsed_schedule)
print("\nOptimized Timetable Schedule via Cuckoo Search:")
print(df_res.sort_values(by="Course ID").to_string(index=False))
```
