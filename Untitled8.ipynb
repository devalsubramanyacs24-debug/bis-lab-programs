```python


import numpy as np
import networkx as nx
import matplotlib.pyplot as plt
import random


def create_network():
    G = nx.DiGraph()


    edges = [
        (0, 1, 4, 1.0), (0, 2, 2, 2.5),
        (1, 3, 5, 1.2), (1, 4, 1, 3.0),
        (2, 1, 1, 1.0), (2, 4, 6, 1.0), (2, 5, 3, 1.5),
        (3, 6, 2, 1.0),
        (4, 3, 2, 1.1), (4, 5, 2, 2.0), (4, 6, 4, 1.5),
        (5, 6, 5, 1.0)
    ]

    for u, v, latency, congestion in edges:
        effective_cost = latency * congestion
        G.add_edge(u, v, latency=latency, congestion=congestion,
                   cost=effective_cost, pheromone=1.0)

    return G

class AntColonyRouting:
    def __init__(self, graph, source, destination, n_ants=10, n_iterations=30, alpha=1.0, beta=2.0, decay=0.1, Q=10):
        self.graph = graph
        self.source = source
        self.destination = destination
        self.n_ants = n_ants
        self.n_iterations = n_iterations
        self.alpha = alpha
        self.beta = beta
        self.decay = decay
        self.Q = Q

    def run(self):
        best_path = None
        best_cost = float('inf')
        cost_history = []

        for iteration in range(self.n_iterations):
            paths = []
            costs = []

            for ant in range(self.n_ants):
                path, cost = self.find_path()
                if path[-1] == self.destination:
                    paths.append((path, cost))
                    costs.append(cost)
                    if cost < best_cost:
                        best_cost = cost
                        best_path = path

            for u, v in self.graph.edges():
                self.graph[u][v]['pheromone'] *= (1 - self.decay)

                if self.graph[u][v]['pheromone'] < 0.01:
                    self.graph[u][v]['pheromone'] = 0.01

            for path, cost in paths:
                for i in range(len(path) - 1):
                    u, v = path[i], path[i+1]
                    self.graph[u][v]['pheromone'] += (self.Q / cost)

            cost_history.append(best_cost if best_cost != float('inf') else None)

        return best_path, best_cost, cost_history

    def find_path(self):
        current_node = self.source
        path = [current_node]
        total_cost = 0
        visited = set([current_node])

        while current_node != self.destination:
            neighbors = list(self.graph.successors(current_node))
            unvisited_neighbors = [n for n in neighbors if n not in visited]

            if not unvisited_neighbors:
                break

            probabilities = []
            for n in unvisited_neighbors:
                edge_data = self.graph[current_node][n]
                pheromone = edge_data['pheromone']
                heuristic = 1.0 / edge_data['cost']

                prob_weight = (pheromone ** self.alpha) * (heuristic ** self.beta)
                probabilities.append(prob_weight)

            sum_weights = sum(probabilities)
            if sum_weights == 0:

                probabilities = [1.0 / len(unvisited_neighbors)] * len(unvisited_neighbors)
            else:
                probabilities = [p / sum_weights for p in probabilities]


            next_node = random.choices(unvisited_neighbors, weights=probabilities, k=1)[0]

            total_cost += self.graph[current_node][next_node]['cost']
            current_node = next_node
            path.append(current_node)
            visited.add(current_node)

        return path, total_cost

if __name__ == "__main__":
    network = create_network()
    source_router = 0
    dest_router = 6

    print(f"--- Running ACO Packet Routing Optimization from Router {source_router} to Router {dest_router} ---")

    aco_router = AntColonyRouting(network, source=source_router, destination=dest_router,
                                  n_ants=15, n_iterations=40, alpha=1.0, beta=2.5, decay=0.15)
    optimal_route, optimal_cost, history = aco_router.run()

    print(f"\nOptimal Routing Path Found: {' -> '.join(map(str, optimal_route))}")
    print(f"Total Link Cost (Latency * Congestion): {optimal_cost:.2f}\n")

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

    ax1.plot(history, marker='o', color='crimson', linewidth=2)
    ax1.set_title("ACO Routing Convergence Curve")
    ax1.set_xlabel("Iteration Loop")
    ax1.set_ylabel("Best Network Cost Found")
    ax1.grid(True, linestyle='--')

    pos = {0: (0, 1), 1: (1, 2), 2: (1, 0), 3: (3, 2), 4: (2, 1), 5: (3, 0), 6: (4, 1)}

    nx.draw_networkx_nodes(network, pos, ax=ax2, node_color='lightgray', node_size=600)
    nx.draw_networkx_labels(network, pos, ax=ax2, font_weight='bold')
    nx.draw_networkx_edges(network, pos, ax=ax2, edgelist=network.edges(), edge_color='gray', width=1, arrows=True, arrowsize=15)

    optimal_edges = [(optimal_route[i], optimal_route[i+1]) for i in range(len(optimal_route)-1)]
    nx.draw_networkx_edges(network, pos, ax=ax2, edgelist=optimal_edges, edge_color='limegreen', width=3.5, arrows=True, arrowsize=20)

    edge_labels = {(u, v): f"{d['cost']:.1f}" for u, v, d in network.edges(data=True)}
    nx.draw_networkx_edge_labels(network, pos, edge_labels=edge_labels, ax=ax2, font_size=9)

    ax2.set_title(f"Optimal Data Packet Path: {'->'.join(map(str, optimal_route))}")
    ax2.axis('off')

    plt.tight_layout()
    plt.show()
```
